```python
import numpy as np
import sympy as sp

def g(x):
    return x**3 + 2*x

# 1. Numerical derivative at x = 2 (use h = 1e-6)
x = 2
h = 1e-6

numerical_derivative = (g(x + h) - g(x - h)) / (2 * h)

print("Numerical derivative at x = 2:")
print(numerical_derivative)

# 2. Symbolic derivative of x**3 + 2*x
x_sym = sp.symbols('x')
g_sym = x_sym**3 + 2*x_sym

symbolic_derivative = sp.diff(g_sym, x_sym)

print("\nSymbolic derivative:")
print(symbolic_derivative)

# 3. Evaluate the symbolic derivative at x = 2 and compare
symbolic_at_2 = symbolic_derivative.subs(x_sym, 2)

print("\nSymbolic derivative at x = 2:")
print(symbolic_at_2)

print("\nAre the numerical and symbolic derivatives approximately equal?")
print(np.isclose(numerical_derivative, float(symbolic_at_2)))
```

```python
import sympy as sp

# Define symbolic variables
x, y = sp.symbols('x y')

# Define the function
f2 = x**2 + 3*x*y + y**2

# 1. Partial derivative with respect to x
print('df/dx =', sp.diff(f2, x))

# 2. Partial derivative with respect to y
print('df/dy =', sp.diff(f2, y))
```

```python
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)

# Evaluate the gradient at the point (x=1, y=2)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)   # points in the steepest-ascent direction
```

```python
import sympy as sp

# Define symbolic variables
x, y = sp.symbols('x y')

# Define the function
h2 = x**2 * y + sp.sin(y)

# 1. dh/dx and dh/dy
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)

# 2. Assemble the gradient list
gradient = [dh_dx, dh_dy]

print("\nGradient =", gradient)

# 3. Evaluate the gradient at (x=2, y=0)
gradient_at_point = [
    expr.subs({x: 2, y: 0})
    for expr in gradient
]

print("\nGradient at (x=2, y=0) =", gradient_at_point)
```

```python
import sympy as sp

# Define symbolic variable
x = sp.symbols('x')

# Define the function
y = sp.exp(x**2 + 1)

# 1. By hand (using the chain rule)
# y = exp(x^2 + 1)
# inner = x^2 + 1
# inner' = 2x
# outer' = exp(inner)

dy_dx_hand = 2*x * sp.exp(x**2 + 1)

print("Derivative by hand:")
print(dy_dx_hand)

# 2. With sp.diff
dy_dx_sympy = sp.diff(y, x)

print("\nDerivative using sp.diff:")
print(dy_dx_sympy)

# 3. Confirm they match
print("\nDo they match?")
print(sp.simplify(dy_dx_hand - dy_dx_sympy) == 0)
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1                 # linear layer 1
h  = np.maximum(0, z1)      # ReLU activation
y_hat = h @ W2              # linear layer 2 (prediction)
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
dy   = 2 * (y_hat - Y) / Y.size      # d loss / d y_hat
dW2  = h.T @ dy                      # d loss / d W2
dh   = dy @ W2.T                     # d loss / d h
dz1  = dh * (z1 > 0)                 # ReLU gradient (1 where z1>0 else 0)
dW1  = X.T @ dz1                     # d loss / d W1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
lr = 0.1
W1 -= lr * dW1               # step downhill
W2 -= lr * dW2

# Recompute the loss after the update
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')

```

```python
import numpy as np

# Data and weights
Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)

Wa = np.random.randn(4, 8) * 0.1     # input -> hidden
Wb = np.random.randn(8, 1) * 0.1     # hidden -> output


# ReLU function
def relu(x):
    return np.maximum(0, x)


# 1. Forward pass
z1 = Xb @ Wa
h = relu(z1)

y_hat = h @ Wb

# Mean squared error loss
loss = np.mean((y_hat - Yb) ** 2)

print("Loss before update:", loss)


# 2. Backward pass

# Derivative of MSE loss with respect to y_hat
dy = (2 / Xb.shape[0]) * (y_hat - Yb)

# Gradient for Wb
dWb = h.T @ dy

# Gradient flowing back to hidden layer
dh = dy @ Wb.T

# Derivative of ReLU
dz1 = dh * (z1 > 0)

# Gradient for Wa
dWa = Xb.T @ dz1


# 3. One gradient-descent step
lr = 0.05

Wa = Wa - lr * dWa
Wb = Wb - lr * dWb


# Forward pass again after the update
z1_after = Xb @ Wa
h_after = relu(z1_after)
y_hat_after = h_after @ Wb

loss_after = np.mean((y_hat_after - Yb) ** 2)

print("Loss after update:", loss_after)
```

```python
xv = 0.0          # starting point
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)        # the derivative
    xv = xv - lr * grad        # step against the gradient
print('Converged x:', round(xv, 3), ' (true minimum = 4)')

```
